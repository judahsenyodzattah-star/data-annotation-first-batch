## 6. Machine learning workflow
A typical supervised-learning workflow:
1. Define the question and target.
2. Inspect and clean the data.
3. Separate features (`X`) from target (`y`).
4. Split data into training and test sets.
5. Train a model using training data.
6. Predict on held-out test data.
7. Evaluate results and document limitations.

This tiny dataset is synthetic for practice; it is not evidence about real students.

````python
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score, confusion_matrix, classification_report,
    roc_curve, roc_auc_score
)

X = df[["hours_studied", "attendance"]]
y = df["passed"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

model = make_pipeline(
    StandardScaler(),
    LogisticRegression(random_state=42)
)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
y_prob = model.predict_proba(X_test)[:, 1]

print("Predictions:", y_pred)
print("Actual:", y_test.to_list())
print("Accuracy:", round(accuracy_score(y_test, y_pred), 3))
````

## 7. Confusion matrix and classification metrics

For binary classification:

- **True positive (TP):** positive case correctly predicted positive.
- **True negative (TN):** negative case correctly predicted negative.
- **False positive (FP):** negative case incorrectly predicted positive.
- **False negative (FN):** positive case incorrectly predicted negative.
- **Precision:** of predicted positives, how many were correct?
- **Recall:** of actual positives, how many did the model find?
- **F1-score:** harmonic mean of precision and recall.
- **Accuracy:** fraction of all predictions that were correct.

A high accuracy score can be misleading when classes are imbalanced.

````python
cm = confusion_matrix(y_test, y_pred, labels=[0, 1])
print("Confusion matrix (rows = actual, columns = predicted):")
print(cm)
print(classification_report(y_test, y_pred, zero_division=0))
````

### Draw the confusion matrix

````python
fig, ax = plt.subplots(figsize=(4.5, 4))
im = ax.imshow(cm)
ax.set_xticks([0, 1], labels=["Predicted 0", "Predicted 1"])
ax.set_yticks([0, 1], labels=["Actual 0", "Actual 1"])
ax.set_title("Confusion matrix")
for i in range(2):
    for j in range(2):
        ax.text(j, i, cm[i, j], ha="center", va="center")
plt.colorbar(im, ax=ax)
plt.tight_layout()
plt.show()
````

## 8. ROC curve and AUC

The ROC curve plots the true-positive rate (recall) against the false-positive rate across classification thresholds. **AUC** summarises ranking performance: 0.5 is roughly random ranking and 1.0 is perfect ranking. On a very small dataset, the estimate is unstable; use a larger, representative test set for meaningful evaluation.

````python
if len(np.unique(y_test)) == 2:
    fpr, tpr, thresholds = roc_curve(y_test, y_prob)
    auc_value = roc_auc_score(y_test, y_prob)
    plt.figure(figsize=(6, 4))
    plt.plot(fpr, tpr, marker=".")
    plt.plot([0, 1], [0, 1], linestyle="--")
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.title(f"ROC curve (AUC = {auc_value:.3f})")
    plt.grid(True)
    plt.show()
    print("AUC:", round(auc_value, 3))
else:
    print("ROC/AUC needs both classes in the test split.")
````

### Experiment with a threshold

````python
# A default classification threshold is usually 0.5.
# Try another threshold and compare false positives vs false negatives.
threshold = 0.5
y_pred_custom = (y_prob >= threshold).astype(int)
print("Threshold:", threshold)
print("Confusion matrix:")
print(confusion_matrix(y_test, y_pred_custom, labels=[0, 1]))
````

## 9. Logistic regression — concept notes

Despite its name, logistic regression is commonly used for **classification**. It estimates a probability for a class, then assigns a class using a threshold.

- **Features:** input variables, such as hours studied and attendance.
- **Target/label:** what the model learns to predict, here `passed`.
- **Training:** model parameters are fitted using training examples.
- **Inference:** the fitted model makes predictions on new examples.
- **Overfitting:** model fits training examples too closely and generalises poorly.
- **Data leakage:** information unavailable at prediction time accidentally enters training.

## 11. Large language models (LLMs)

A **large language model (LLM)** is trained on large amounts of text to predict/generate token sequences. It can summarise, explain, classify, translate and help with code, but can also produce confident errors.

Key terms:
- **Token:** a unit of text processed by the model.
- **Prompt:** the instruction and context supplied to the model.
- **Inference:** using a trained model to generate an output.
- **Temperature:** a generation setting that generally affects randomness.
- **Context window:** the input/output context the model can process in one request.
- **Hallucination:** plausible-sounding output that is incorrect or unsupported.

## 13. Retrieval-augmented generation (RAG)

**RAG** combines information retrieval with an LLM response. Instead of relying only on what the model learned during training, a RAG system retrieves relevant passages from a supplied knowledge source and includes them in the prompt.

Typical flow:
1. Collect trusted documents.
2. Split documents into chunks.
3. Convert chunks into embeddings (numeric representations).
4. Store embeddings in a vector index/database.
5. Embed the user's query and retrieve relevant chunks.
6. Give the retrieved context and question to the LLM.
7. Check that the answer is supported by the retrieved sources.

RAG can improve grounding, but it does not guarantee correctness. Retrieval quality, source quality, chunking, permissions and citations all matter.

````python
# Tiny retrieval illustration without embeddings or an LLM.
documents = [
    "Precision is the fraction of predicted positive cases that are correct.",
    "Recall is the fraction of actual positive cases that are identified.",
    "AUC summarises the ROC curve across thresholds.",
    "A confusion matrix compares actual and predicted classes.",
]
query = "How do I understand recall?"
query_words = set(query.lower().replace("?", "").split())

def simple_keyword_retrieval(query, documents):
    words = set(query.lower().replace("?", "").split())
    scored = []
    for doc in documents:
        doc_words = set(doc.lower().replace(".", "").split())
        scored.append((len(words & doc_words), doc))
    return sorted(scored, reverse=True)

for score, doc in simple_keyword_retrieval(query, documents)[:2]:
    print(score, "—", doc)

print("\nNote: this is only a keyword-overlap demo, not a production RAG system.")
````

## 14. Model testing and responsible AI

Before trusting a model, ask:
- Does the test set represent the real use case?
- Are classes imbalanced?
- Which errors are most harmful?
- Is there data leakage?
- Can the result be reproduced from recorded settings?
- Does the dataset have appropriate consent, rights and privacy protections?
- Could performance differ across groups or conditions?
- Is a human review step needed?

For face detection, distinguish **detecting a face** (finding a bounding box) from **identifying a person** (recognition). They are different tasks with different privacy implications.
